```python
data = {'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Bob', 'Alice'],

'Age': [25, 30, 35, None, 28, 30, None],

'City': ['New York', 'London', 'Paris','New York', 'London', 'Paris', 'Houston', ]}

df = pd.DataFrame(data)

print(df)
```

```python
over_25 = df[df['Age'] > 25] 

print(over_25)
```

```python
df
```

```python
df['City'] = df['City'].replace('New York', 'NYC')

print(df)
```

```python
#df_filled = df.fillna(0)
#df_filled
# Drop rows with any missing values

#df_dropped_na = df.dropna()
#df_dropped_na
# Handling missing values (fill with mean)

df['Age'] = df['Age'].fillna(df['Age'].mean())

print(df)
```

```python
df_grouped = df.groupby('City')['Age'].mean()

print(df_grouped)
```

```python
df2 = pd.DataFrame({'Name': ['Alice', 'Bob'], 'Salary': [50000, 60000]})

df_merged = pd.merge(df, df2, on='Name')

print(df_merged)
```

```python
df3 = pd.DataFrame({'Name': ['David'], 'Age': [40], 'City': ['Berlin']})

df_concat = pd.concat([df, df3])

print(df_concat)
```

```python
df.drop_duplicates(inplace=True)
print(df)
```
