V. VISUALIZE

1. What is the gender distribution of students?

```python
sns.countplot(x='gender', data=data)
plt.title('Gender Count')
plt.savefig('gender_count.png')
plt.show()
```

2. What is the average math score for males vs females?

```python
sns.barplot(x='gender', y='math score', data=data)
plt.title('Average Math Score by Gender')
plt.savefig('avg_math_by_gender.png')
plt.show()
```

3. What is the average reading score for males vs females?

```python
sns.barplot(x='gender', y='reading score', data=data)
plt.title('Average Reading Score by Gender')
plt.savefig('avg_reading_by_gender.png')
plt.show()


```

4. What is the average writing score for males vs females?

```python
sns.barplot(x='gender', y='writing score', data=data)
plt.title('Average Writing Score by Gender')
plt.savefig('avg_writing_by_gender.png')
plt.show()
```

5. How does parental level of education affect average math scores?

```python
sns.barplot(x='parental level of education', y='math score', data=data, estimator='mean')
plt.title('Average Math Score by Parental Education')
plt.xticks(rotation=45)
plt.savefig('avg_math_by_parent_edu.png')
plt.show()
```

6. Do students who complete the test preparation course have higher average math scores?

```python
sns.barplot(x='test preparation course', y='math score', data=data, estimator='mean')
plt.title('Average Math Score by Test Preparation')
plt.savefig('avg_math_by_test_prep.png')
plt.show()
```

7. What is the distribution of students by race/ethnicity?

```python
sns.countplot(x='race/ethnicity', data=data)
plt.title('Gender Count')
plt.savefig('gender_count.png')
plt.show()
```

8. What is the distribution of parental levels of education?

```python
sns.countplot(x='parental level of education', data=data)
plt.title('Distribution of Parental Education Level')
plt.xticks(rotation=45)
plt.show()
```

9. How many students have standard vs free/reduced lunch?

```python
sns.countplot(x='lunch', data=data)
plt.title('Distribution of Parental Education Level')
plt.xticks(rotation=45)
plt.show()
```

```python
lunch_counts = data['lunch'].value_counts()
lunch_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Lunch Type Distribution')
plt.show()
```

10. Do students who completed the test preparation course score higher on average?

```python
sns.barplot(x='test preparation course', y='math score', data=data)
plt.title('Average Math Score by Test Preparation')
plt.show()
```

11. How does parental education level affect average math/reading/writing scores?

```python
sns.barplot(x='parental level of education', y='math score', data=data)
plt.title('Average Math Score by Parental Education')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(x='parental level of education', y='reading score', data=data)
plt.title('Average Reading Score by Parental Education')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(x='parental level of education', y='writing score', data=data)
plt.title('Average Writing Score by Parental Education')
plt.xticks(rotation=45)
plt.show()
```

12. How does lunch type affect scores?

```python
sns.barplot(x='lunch', y='math score', data=data)
plt.title('Average Math Score by Lunch Type')
plt.show()

```

```python
sns.barplot(x='lunch', y='reading score', data=data)
plt.title('Average Reading Score by Lunch Type')
plt.show()

```

```python
sns.barplot(x='lunch', y='writing score', data=data)
plt.title('Average Writing Score by Lunch Type')
plt.show()
```

13. What is the relationship between math and reading scores?

```python
sns.scatterplot(x='math score', y='reading score', data=data)
plt.title('Math vs Reading Scores')
plt.show()
```

14. What is the relationship between math and reading scores?

```python
sns.scatterplot(x='reading score', y='writing score', data=data)
plt.title('Reading vs Writing Scores')
plt.show()
```

15. Do average scores differ between male and female students?

```python
sns.barplot(x='gender', y='math score', data=data)
plt.title('Average Math Score by Gender')
plt.show()
```

```python
sns.barplot(x='gender', y='reading score', data=data)
plt.title('Average Reading Score by Gender')
plt.show()
```

```python
sns.barplot(x='gender', y='writing score', data=data)
plt.title('Average Writing Score by Gender')
plt.show()
```

15. What is the average total score (math + reading + writing) by gender?


```python
data['total score'] = data['math score'] + data['reading score'] + data['writing score']
sns.barplot(x='gender', y='total score', data=data)
plt.title('Average Total Score by Gender')
plt.show()

```
